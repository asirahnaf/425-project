Genre/tag distributions, mel preview, graph stats. Run from repo root (`ROOT` auto-detects).

```python
import sys
from pathlib import Path
import json
from collections import Counter
ROOT = Path.cwd()
if not (ROOT/'src').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
print('ROOT:', ROOT)
gtzan = list((ROOT/'data/raw/gtzan').rglob('*.wav'))
print('GTZAN wavs:', len(gtzan))
print(Counter([p.parent.name for p in gtzan]))
print('npz:', len(list((ROOT/'data/processed').glob('*.npz'))))
print('graphs .pt:', len(list((ROOT/'data/processed/graphs').glob('*.pt'))))
import csv
mc = list(csv.DictReader(open(ROOT/'data/raw/musiccaps-public.csv', encoding='utf-8')))
print('MusicCaps rows:', len(mc), 'cols:', list(mc[0].keys())[:6])
print('caption ex:', mc[0]['caption'][:160] if 'caption' in mc[0] else list(mc[0].values())[0][:160])
```

```python
import numpy as np, matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from collections import Counter
from pathlib import Path
counts = Counter([p.parent.name for p in (ROOT/'data/raw/gtzan').rglob('*.wav')])
genres = sorted(counts)
fig, ax = plt.subplots(figsize=(8,3))
ax.bar(genres, [counts[g] for g in genres])
ax.set_title('GTZAN per-genre counts (target 100/genre)')
ax.set_ylabel('tracks'); plt.xticks(rotation=30)
fig.tight_layout()
fig.savefig(ROOT/'results/plots/eda_genre_hist.png', dpi=150)
print('saved results/plots/eda_genre_hist.png')
npz = sorted((ROOT/'data/processed').glob('*.npz'))[0]
d = np.load(str(npz))
print(npz.name, 'mel', d['mel'].shape, 'chroma', d['chroma'].shape, 'mfcc', d['mfcc'].shape)
fig2, ax2 = plt.subplots(figsize=(8,3))
ax2.imshow(d['mel'][:, :256], aspect='auto', origin='lower')
ax2.set_title(f"log-mel ex {npz.name} [128 x T]")
fig2.tight_layout()
fig2.savefig(ROOT/'results/plots/eda_mel_ex.png', dpi=150)
print('saved results/plots/eda_mel_ex.png')
```
